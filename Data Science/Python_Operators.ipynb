```python
# Comparison Operators
# These operators are used to compare values, They return value True or False.
# Equal To(==)
# Description: Checks if two values are equal.
#Special Note: Commonly used in conditional and validations.

marks = 75
print('Passed == 75:', marks == 75)
```

```python
# Not Equal (!=)
# Description: Checks if two values are not equal.
# Special Note: Used for exclusion lotic.

print('Absent != 0:', marks != 0)
```

```python
# Greater Than (>)
# Description: Checks if the left value is greter thna the right
# Special Note: Useful in grading, thersholds.

marks = 71
print('Grade A:', marks > 70)
```

```python
# Less Than (<)
# Description: Checks if the left value is less than the right
# Special Note: Used in limit checks or age filtering

marks = 39
print('Needs Improvement:', marks < 40)
```

```python
# Greter Than or Equal To (>=)
# Description: Checks if the left is greater or equal.
# Special Note: Used in pss/fail conditions.

marks = 60
print('First Class:', marks >= 60)
```

```python
# Less Than or Equal To (<=)
# Description: Checks if the left is less or equal.
# Special Note: Used in eligibility logic.

marks = 45
print('Below Average:', marks <= 50)
```

```python
# Logical Operators
# These Operators are used to combine conditionals statements.
# Logical AND (and)
# Description: Result True if both conditions are treu.
# Special Note: Used in multi-condition filtering.

bp = 135
sugar = 180
print('Critical:', bp > 130 and sugar > 150)
```

```python
# Logical OR (or)
# Description: Returns True of at least one conditions is true.
# Special Note: Used in decision trees.

bp = 170
sugar = 130
print('Requires Checkup:', bp > 30 or sugar > 150)
```

```python
# Logical NOT (not)
# Description: Inverts the Boolean value.
# Special Note: Used to toggle logic.

healthy = False
print('Not Health:' , not healthy)
```

```python
# Assignment Operators
#These are used to assign values to variables and modify them using various arithmetic opeartions
# Assign (=)
# Description: Assigns the value on the right to the variable on the left
# Special Note: Most basic form of assignment

stock = 100
print('Initial stock:', stock)
```

```python
# Add and Assign (+ =)
# Description: Adds and assigns the rsult to the variable.
# Special Note: Useful for cumulative totals

stock = 100
stock += 20
print('Updated stock:', stock)
```

```python
# Subtract and Assign (-)
# Description: Subracts and assigns result to the variable.
# Special Note: Used in decrementing inventory.

stock = 120
stock -= 30
print('Remaining stock:', stock)
```

```python
# Multiply and Assign (*=)
# Description: Multiplies and assigns result to the variable
# Special Note: Used in progressive increases.

demand = 10
demand *= 3
print('Total Demand:', demand)
```

```python
# Divide and Assign (/=)
# Description: Divides and assigns result to the variable
# Special Note: Used in averaging or scalling

weight = 300
weight /= 3
print('Average Weight:', weight)
```

```python
# Modulus and Assign (%=)
# Description: Applies modulus and assigns result.
# Special Note: Useful for wrapping or cycles.

count = 17
count %= 5
print('Remainig Count:', count)
```

```python
# Bitwise Operators
# These operators work on bits and perform bit-by-bit operations.
# AND (&)
# Description: Perform bitwise AND.
# Special Note: Used in IoT for permission flags or staatus registers.

# Imaging we have two simple numbers 
a = 5   # In binary: 0101
b = 5   # In binary: 0011

# Bitwise AND compares each bit position
# 0101 (5)
# 0011 (3)
# ________
# 0001 (1) only the last bit is 1 in both numbers

result = a & b
print(result) # Output: 1
```
